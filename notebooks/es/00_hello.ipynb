## 1. ¿De verdad estamos ejecutando R?

Esto debe imprimir una versión de R. Si da error, o imprime algo sobre Python, el
runtime está equivocado: ve a **Entorno de ejecución → Cambiar tipo de entorno de
ejecución** y elige **R**.

```r
cat(R.version.string, "\n")
cat("platform:", R.version$platform, "\n")
```

## 2. ¿En qué máquina estamos?

Colab te da una máquina Linux temporal en un centro de datos de Google. Nada de
esto corre en tu laptop. Esta celda reporta de qué Linux se trata.

Nota cómo llamamos a un comando de la terminal: desde R usamos `system()`, no `!`.

```r
cat(system("cat /etc/os-release 2>&1", intern = TRUE), sep = "\n")
```

## 3. ¿Dónde viven los paquetes de R?

Los notebooks del taller instalarán paquetes en esta carpeta. Se borra cuando el
runtime se apaga, lo cual es normal y esperado.

```r
cat("library paths:\n")
cat(paste0("  ", .libPaths()), sep = "\n")
cat("\npackages already installed:", nrow(installed.packages()), "\n")
```

## 4. ¿Podemos dibujar un gráfico?

Las figuras son el punto de la mayor parte de este taller, así que confirmemos que
aparece una. Las opciones `repr.plot` fijan el tamaño de cada figura del notebook;
las ponemos una vez al inicio de cada notebook.

```r
options(repr.plot.width = 10, repr.plot.height = 7)

plot(
  iris$Petal.Length, iris$Petal.Width,
  col  = iris$Species,
  pch  = 19,
  xlab = "petal length",
  ylab = "petal width",
  main = "If you can see this, plotting works"
)
legend("topleft", legend = levels(iris$Species), col = 1:3, pch = 19)
```
